```python
import pandas as pd
from pathlib import Path

# Project directory
BASE_DIR = Path.cwd().parent

# Input dataset
DATA_PATH = BASE_DIR / "data" / "processed" / "model_data.csv"

# Output testing dataset
OUTPUT_PATH = BASE_DIR / "data" / "processed" / "test_data.csv"

# Load complete dataset
df = pd.read_csv(DATA_PATH)

# Extract testing split
test_df = df[df["split"] == "test"].copy()

# Save as separate CSV
test_df.to_csv(OUTPUT_PATH, index=False)

print("Testing dataset created successfully!")
print("Shape:", test_df.shape)
print("Saved at:", OUTPUT_PATH)
```
